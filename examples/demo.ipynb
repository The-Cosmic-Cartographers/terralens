# TerraLens package demo

Runs top to bottom with no network. Every number comes from the committed
series under `data/processed/` and the same statistics the app uses.

````python
import terralens as tl

tl.list_regions()
````

````python
ireland = tl.ireland
print(ireland.info())
````

`info()` reads the region registry only — no parquet is opened until a
variable is accessed.

````python
ireland.summary()
````

````python
trend = ireland.temperature.trend()
trend
````

````python
trend.plot()
````

````python
comparison = tl.compare("ireland", "sahel", variables=["temperature"])
print(comparison)
````

````python
comparison.table
````

## Adding a region or a variable

A region is one entry in `pipeline/regions.json`:

```json
{"region_id": "my_box", "name": "My box",
 "west": 8.0, "south": 45.0, "east": 10.0, "north": 47.0, "scope": "all"}
```

It appears in `tl.list_regions()` as soon as its series are processed.
A variable is one small subclass in `terralens/variables/` that names its
dataset id; the statistics are inherited.
